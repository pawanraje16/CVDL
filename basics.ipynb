```python
img=cv2.imread('cat.jpg')
```

```python
#Display the image
plt.imshow(img)
plt.show()
```

```python
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img_rgb)
plt.show()
```

```python
img_rgb.shape
```

```python
type(img_rgb)
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray)
plt.show()
```

```python
plt.imshow(gray,cmap='gray')
plt.show()
```

```python
gray.shape
```

```python
type(gray)
```

```python
#Display the image
plt.imshow(img_rgb[:,:,0],cmap='gray')
plt.show()
```

```python
#Display the image
plt.imshow(img_rgb[:,:,1],cmap='gray')
plt.show()
```

```python
plt.imshow(img_rgb[:,:,2],cmap='gray')
plt.show()
```

```python
# Create a figure
fig=plt.figure(figsize=(8,6)) # specify figure size in inches
#Add a single subplot to the figure
ax=fig.add_subplot(111) # (1 row , 1 column , 1st subplot)
# Display the image
ax.imshow(img_rgb)
ax.axis('off') # Turn off axis Labels and ticks
ax.set_title("Image Display") # Set a title for the subplot
# Show the figure
plt.show()
```

```python
cv2.line(img_rgb,(0,600),(600,600),(255,0,23),10)
cv2.rectangle(img_rgb,(100,100),(160,160),(0,255,0),3)
cv2.circle(img_rgb,(1000,670),40,(34,12,255),-1)
plt.imshow(img_rgb)
plt.show()
```
